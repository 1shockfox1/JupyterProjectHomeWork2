11.1 средний

```python
class Date:

    def __init__(self):
        self.day = 1
        self.month = 1
        self.year = 2000


    def __init__(self, day=1, month=1, year=2000):
        self.day = day
        self.month = month
        self.year = year


    def __del__(self):
        print(f"Объект Date({self.day}.{self.month}.{self.year}) уничтожен")

# Метод 1
    def is_leap(self):
        return self.year % 4 == 0

# Метод 2
    def add_days(self, n=5):
        days_in_month = [31, 28, 31, 30, 31, 30,
                         31, 31, 30, 31, 30, 31]
        if self.is_leap():
            days_in_month[1] = 29

        self.day += n
        while self.day > days_in_month[self.month - 1]:
            self.day -= days_in_month[self.month - 1]
            self.month += 1
            if self.month > 12:
                self.month = 1
                self.year += 1

                days_in_month[1] = 29 if self.is_leap() else 28


    def __str__(self):
        return f"{self.day:02d}.{self.month:02d}.{self.year}"



d1 = Date(15, 3, 2024)

print("Дата:", d1)

print("\n Високосный ли год? ")
print(f"{d1} → год {d1.year} високосный? {d1.is_leap()}")

print("\nУвеличение даты на 5 дней ")
d1.add_days(5)
print("Дата  после +5 дней:", d1)

d3 = Date(28, 12, 2023)
print("До:", d3)
d3.add_days(5)
print("После +5 дней:", d3)


```

11.2 средний

```python
class EuroConverter:
    def __init__(self, rate):
        self.rate = rate

    def to_euro(self, bills):
        return sum(bills) / self.rate

    def __str__(self):
        return f"1 € = {self.rate} грн"


conv = EuroConverter(45.0)
bills = [100, 200, 500, 1000]

print(conv)
print(f"{sum(bills)} грн = {conv.to_euro(bills):.2f} €")
```

11.3 обычный


```python
class Computer:
    def __init__(self, cpu, mhz, ram):
        self.cpu = cpu      # процессор
        self.mhz = mhz      # МГц
        self.ram = ram      # Мб

    def Q(self):            # качество 1-го уровня
        return 0.1 * self.mhz + self.ram

    def __str__(self):
        return f"{self.cpu}, {self.mhz} МГц, {self.ram} Мб"


class ComputerP(Computer):
    def __init__(self, cpu, mhz, ram, hdd):
        super().__init__(cpu, mhz, ram)
        self.hdd = hdd

    def Q(self):
        return super().Q() + 0.5 * self.hdd

    def __str__(self):
        return super().__str__() + f", HDD {self.hdd} Гб"



c1 = Computer("Intel i5", 3000, 8192)
c2 = ComputerP("Intel i7", 3600, 16384, 512)

print(c1, "→ Q =", c1.Q())
print(c2, "→ Qp =", c2.Q())
```
